```python
from ultralytics import YOLO
import shutil

model = YOLO('yolov8n.pt')

results = model.train(
    data='/content/drive/MyDrive/Mini Project/aaadhaaar/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16
)

# Path where Ultralytics saved the best weights
best_path = results.save_dir / 'weights' / 'best.pt'

# Copy/rename it to bestmodel.pt
dest_path = '/content/drive/MyDrive/Mini Project/aaadhaaar/bestmodel.pt'
shutil.copy(str(best_path), dest_path)

print(f"Best model saved as: {best_path}")
```

```python
import matplotlib.pyplot as plt
from ultralytics import YOLO
import glob
import math

# Load your trained model
model_path = '/content/drive/MyDrive/Mini Project/aaadhaaar/bestmodel.pt'
model = YOLO(model_path)

# Path to test images
test_dir = '/content/drive/MyDrive/Mini Project/aaadhaaar/images/test'
test_images = glob.glob(test_dir + '/*.jpg') + glob.glob(test_dir + '/*.png') + glob.glob(test_dir + '/*.jpeg')

print(f"Found {len(test_images)} test images")




# Run inference on all images
results_list = [model(img_path) for img_path in test_images]

# Plot all images in a grid
n = len(test_images)
cols = 3
rows = math.ceil(n / cols)

fig, axes = plt.subplots(rows, cols, figsize=(5*cols, 5*rows))
axes = axes.flatten() if n > 1 else [axes]

for i, (img_path, results) in enumerate(zip(test_images, results_list)):
    annotated = results[0].plot()[..., ::-1]  # BGR -> RGB
    axes[i].imshow(annotated)
    axes[i].set_title(img_path.split('/')[-1], fontsize=8)
    axes[i].axis('off')

# Hide unused subplots
for j in range(i+1, len(axes)):
    axes[j].axis('off')
    plt.tight_layout()
    plt.show()
```

```python
from pathlib import Path
import shutil

# Original YOLO project folder
source_project = Path(
    "/content/drive/MyDrive/Mini Project/aaadhaaar"
)

# Original Colab notebook
source_notebook = Path(
    "/content/drive/MyDrive/Colab Notebooks/adhaar_detect_using_yolo.ipynb"
)

# Destination folder
github_project = Path(
    "/content/aadhaar-detection-yolo"
)

# Verify source files
assert source_project.is_dir(), "Project folder not found!"
assert source_notebook.is_file(), "Notebook not found!"

# Copy entire project including all subfolders
shutil.copytree(
    source_project,
    github_project,
    dirs_exist_ok=True
)

# Copy notebook into project
shutil.copy2(
    source_notebook,
    github_project / source_notebook.name
)

print("COMPLETE PROJECT COPIED SUCCESSFULLY!")

# Show project structure
for root, dirs, files in __import__("os").walk(github_project):
    level = root.replace(str(github_project), "").count(__import__("os").sep)
    indent = " " * 4 * level

    print(f"{indent}{Path(root).name}/")

    for file in files:
        print(f"{indent}    {file}")
```
